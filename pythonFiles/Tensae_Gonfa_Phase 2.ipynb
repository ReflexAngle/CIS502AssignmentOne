## FUNCTIONS (Bradley Petersen)

```python
# Create a function that uses sepal and petal lengths to classify flowers as Small, Medium, or Large.

def classify_flower(sepal_length, petal_length):
    if sepal_length < 5 and petal_length < 2:
        return "Small"
    elif sepal_length < 6 and petal_length < 5:
        return "Medium"
    else:
        return "Large"

# Test the function with different measurements to make sure it returns the correct categories.

print(classify_flower(4.5, 1.4))
print(classify_flower(5.5, 4.0))
print(classify_flower(6.5, 5.5))
```
