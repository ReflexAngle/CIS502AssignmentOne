# Phase 3 - File Handling & Data Cleaning

## 3.1 File I/O using Python's Built-In `csv` Module (Timur Zhomartov)

```python
import csv
import pandas as pd
input_filename = "../DataSet/iris.data"
```

```python
rows = []
with open(input_filename, mode="r", encoding="utf-8") as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        if row: 
            rows.append(row)
row_count = len(rows)
```

```python
petal_lengths = [float(row[2]) for row in rows]
min_petal_length = min(petal_lengths)
max_petal_length = max(petal_lengths)
avg_petal_length = sum(petal_lengths) / row_count
```

```python
with open("summary.txt", mode="w", encoding="utf-8") as summary_file:
    summary_file.write(f"Dataset Summary\n")
    summary_file.write(f"----------------\n")
    summary_file.write(f"Total Rows: {row_count}\n")
    summary_file.write(f"Min Petal Length: {min_petal_length}\n")
    summary_file.write(f"Max Petal Length: {max_petal_length}\n") 
    summary_file.write(f"Average Petal Length: {avg_petal_length:.2f}\n")

print("summary.txt successfully created!")
```

## 3.2 Data Inspection (Tensae Gonfa)

In this section, we use pandas to load the Iris dataset and take a closer look at the data. We check the column names, data types, and missing values to see if anything needs to be fixed before cleaning.

```python
import pandas as pd

# Add column names since the original file has none
column_names = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

# Load the Iris dataset
df = pd.read_csv("../DataSet/iris.data", header=None, names=column_names)

# Show the first five rows
print("First five rows of the dataset:")
display(df.head())
```

```python
# Check the column names
print("Column Names:")
print(df.columns.tolist())

# Check the data type of each column
print("\nData Types:")
print(df.dtypes)
```

```python
# Check for missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Check the number of rows and columns
print("\nDataset Shape:")
print(df.shape)
```

### 3.2 Data Inspection Findings

The Iris dataset contains 150 rows and 5 columns. Four columns contain numerical measurements, while the species column identifies the flower type.

We found no missing values, and the data types are appropriate for the information stored in each column. The dataset appears to be in good condition for cleaning and further analysis.
