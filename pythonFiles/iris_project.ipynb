```python
#Bradley Petersen
# return a size label returns, doesn't print, so the label is reusable
def classify_flower(sepal_length, petal_length):
    if petal_length < 2:
        return "small"
    elif petal_length <= 5:
        return "medium"
    else:
        return "large"

print(classify_flower(5.1, 1.4))   # small
print(classify_flower(6.0, 4.5))   # medium
print(classify_flower(6.5, 5.8))   # large
```

```python
# Tensae Gonfa
def classify_flower(sepal_length, petal_length):
    if sepal_length < 5 and petal_length < 2:
        return "Small"
    elif sepal_length < 6 and petal_length < 5:
        return "Medium"
    else:
        return "Large"

print(classify_flower(4.5, 1.4))
print(classify_flower(5.5, 4.0))
print(classify_flower(6.5, 5.5))
```
