### Count Flowers by Species - Tensae Gonfa

```python
import pandas as pd
```

```python
# Load the cleaned Iris dataset from Phase 3
df = pd.read_csv("clean_iris.csv")

# Verify the dataset
print("Dataset Shape:", df.shape)

# Create an empty dictionary to store species counts
species_counts = {}

# Count the flowers of each species using a for loop
for species in df["species"]:
    if species in species_counts:
        species_counts[species] += 1
    else:
        species_counts[species] = 1

# Display the counts for each species
print("\nFlower Counts by Species:")
for species, count in species_counts.items():
    print(species, ":", count)

# Verify the total number of flowers
print("\nTotal Flowers:", sum(species_counts.values()))
```

### Categorize Flowers Using If/Else - Tensae Gonfa

```python
# Create an empty list to store petal categories
petal_categories = []

# Categorize each flower based on its petal length
for length in df["petal_length"]:
    if length < 2:
        petal_categories.append("Short Petal")
    elif length <= 5:
        petal_categories.append("Medium Petal")
    else:
        petal_categories.append("Long Petal")

# Add the categories as a new column
df["petal_category"] = petal_categories

# Display the first 10 rows to verify the categories
print(df[["species", "petal_length", "petal_category"]].head(10))

# Count how many flowers belong to each category
print("\nPetal Category Counts:")
print(df["petal_category"].value_counts())

# Verify the total number of categorized flowers
print("\nTotal Categorized Flowers:", len(petal_categories))
```

### Export Categorized Flowers to CSV - Tensae Gonfa

```python
# Save the categorized Iris dataset to a CSV file
df.to_csv("iris_classes.csv", index=False)

# Read the saved CSV file to verify it
iris_classes = pd.read_csv("iris_classes.csv")

# Display the first 10 rows
print("Categorized Iris Dataset:")
print(iris_classes.head(10))

# Verify the number of rows and columns
print("\nDataset Shape:", iris_classes.shape)

# Verify the petal category counts
print("\nPetal Category Counts:")
print(iris_classes["petal_category"].value_counts())

# Check for missing values
print("\nMissing Values:")
print(iris_classes.isnull().sum())
```
