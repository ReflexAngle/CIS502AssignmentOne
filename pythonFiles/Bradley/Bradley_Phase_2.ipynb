```python
def larget_shotest_petal():
    target_length = 1.5
    # calling the column thats being used
    length_column = data['petal_length']
    # stores metadata aka the first row in the data set 
    largest_pettle = length_column[0]

    # loop through every column of the data set
    for i in length_column:
        # uses the colum name to get the length column
        if i >= target_length:
            print(f'{i} large pettle')
        else:
            print(f'{i} small pettle')

        # seeing what is the largst and smallest variable the prining it
        if i > largest_pettle:
            largest_pettle = i

    print(f'the largest pettle is: {largest_pettle}')
        
larget_shotest_petal()
   
```
