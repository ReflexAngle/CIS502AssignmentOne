```python
import csv
import pandas as pd
input_filename = "iris.data"
rows = []
with open(input_filename, mode="r", encoding="utf-8") as file:
  csv_reader = csv.reader(file)
  for row in csv_reader:
    if row:
        rows.append(row)
row_count = len(rows)

petal_lengths = [float(row[2]) for row in rows]
min_petal_length = min(petal_lengths)
max_petal_length = max(petal_lengths)
avg_petal_length = sum(petal_lengths) / row_count

with open("summary.txt", mode="w", encoding="utf-8") as summary_file:
    summary_file.write(f"Dataset Summary\n")
    summary_file.write(f"----------------\n")
    summary_file.write(f"Total Rows: {row_count}\n")
    summary_file.write(f"Min Petal Length: {min_petal_length}\n")
    summary_file.write(f"Max Petal Length: {max_petal_length}\n")
    summary_file.write(f"Average Petal Length: {avg_petal_length:.2f}\n")

print("summary.txt successfully created!")
```

```python
import pandas as pd

# Load the cleaned Iris dataset from Phase 3
df = pd.read_csv("clean_iris.csv")

# Verify the dataset shape
print("Dataset Shape:", df.shape)

# Calculate mean, median, min, and max for sepal length and petal length
print("Descriptive Statistics for Sepal and Petal Lengths:")
stats_summary = {
    "Statistic": ["Mean", "Median", "Min", "Max"],
    "Sepal Length": [
        df["sepal_length"].mean(),
        df["sepal_length"].median(),
        df["sepal_length"].min(),
        df["sepal_length"].max()
    ],
    "Petal Length": [
        df["petal_length"].mean(),
        df["petal_length"].median(),
        df["petal_length"].min(),
        df["petal_length"].max()
    ]
}

stats_df = pd.DataFrame(stats_summary)
print(stats_df.to_string(index=False))

# Group by species and compute averages for all numerical measurement
print("\nAverage Measurements by Species:")
species_averages = df.groupby("species").mean()
print(species_averages)
```
